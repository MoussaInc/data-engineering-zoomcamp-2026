```python
def csv_to_parquet(spark, taxi_types, year, schema):

    errors = []

    for taxi in taxi_types:
        for month in range(1, 13):

            print(f"processing data for {taxi}/{year}/{month:02d}")

            input_file = f"{gs_bucket}raw/{taxi}/{year}/{taxi}_tripdata_{year}-{month:02d}.csv.gz"

            output_path = f"{gs_bucket}pq/{taxi}/{year}/{month:02d}/"

            try:
                df = (
                    spark.read
                    .option("header", "true")
                    .schema(schema)
                    .csv(input_file)
                )

                (
                    df.repartition(2)
                      .write
                      .mode("overwrite")
                      .parquet(output_path)
                )

                print(f"saved to {output_path}")

            except Exception as e:

                print(f"erreur lors du traitement: {e}")
                errors.append((taxi, year, month, str(e)))

    if errors:
        print(f"\n{len(errors)} erreur(s):")

        for taxi, year, m, msg in errors:
            print(f"{taxi}/{year}/{m:02d}: {msg}")
```

```python
csv_to_parquet(
    spark,
    taxi_types=["yellow"],
    year=2020,
    schema=yellow_schema
)
```

```python
csv_to_parquet(
    spark,
    taxi_types=["yellow"],
    year=2021,
    schema=yellow_schema
) 
```
